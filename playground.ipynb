```python
from transformers import RobertaTokenizer

# Load the pretrained tokenizer
tokenizer = RobertaTokenizer.from_pretrained('roberta-base')

# Define the sentence
sentence = "The answer to life the universe and everything"

# Tokenize the sentence
tokens = tokenizer.tokenize(sentence)

masked_tokens = [
    "<mask>" if token == "ĠLife" else token for token in tokens
]

# Encode the tokens to get token IDs
encoded_input = tokenizer.encode(sentence, return_tensors='pt')
masked_input = tokenizer.encode(" ".join(masked_tokens), return_tensors='pt')

# Print the results
print("Original Tokens:", tokens)
print("Masked Tokens:", masked_tokens)
print("Original Token IDs:", encoded_input)
print("Masked Token IDs:", masked_input)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV einlesen
df = pd.read_csv("cleaned_dataset_metrics.csv", index_col="corr_num_label")

# Daten: Hauptkategorie (main_category) und Anzahl der Fälle (count)
categories = df['main_category']
counts = df['count']
# Prozentuale Verteilung berechnen
percentages = counts / counts.sum() * 100

# Grafikgröße anpassen
plt.figure(figsize=(12, 6))

# Balkendiagramm mit getauschten Achsen und logarithmischer Skala
plt.barh(categories, counts, color='skyblue', edgecolor='black')
plt.xscale('log')  # Logarithmic scale for the x-axis
plt.xlabel('Number of Cases (Log Scale)')
plt.ylabel('Categories')
# plt.title('Distribution of Cases Across Classes (Logarithmic Scale)')
plt.tight_layout()

# Grafik speichern
plt.savefig("logarithmic_class_distribution.png", dpi=300, bbox_inches='tight')

# Grafik anzeigen
plt.show()

plt.savefig("images/logarithmic_class_distribution.png", dpi=300, bbox_inches='tight')
```
